#### 1.1. **(1,5 балла)** Реализация класса `Ingredient`

Класс `Ingredient` представляет собой базовую единицу рецепта – отдельный продукт или компонент блюда (мука, яйца, масло и т.д.). Он нужен для того, чтобы хранить информацию о каждом продукте, его количестве и единицах измерения.

Этот класс должен иметь следующую структуру:
1. Метод `__init__`, инициализирующий три атрибута:
    - `name` – название, строка
    - `quantity` – количество, float
    - `unit` – единица измерения, строка (г, кг, мл, шт)
2. Атрибут `quantity` должен быть реализован через `@property`:
    - геттер возвращает значение
    - сеттер проверяет, что значение положительное; если нет – выбрасывает `ValueError` с сообщением `"Количество должно быть положительным"`
    - сеттер приводит значение к `float`
3. Метод `__str__` – возвращает строку вида: "Мука: 500.0 г".
4. Метод `__repr__` – возвращает строку вида: `"Ingredient('Мука', 500.0, 'г')"`
5. Метод `__eq__` – два ингредиента считаются равными, если у них совпадают `name` и `unit` (количество не важно). Поскольку в рецепте не должно быть двух записей одного ингредиента, такие записи в дальнейшем нужно будет объединять

````python
class Ingredient:
    
    def __init__(self, name: str, quantity: float, unit: str) -> None:
        '''Инициализация ингредиента.
        
        Args:
            name: название ингредиента
            quantity: количество
            unit: единица измерения (г, кг, мл, шт)
        '''
        self.name = name
        self.unit = unit
        self.quantity = quantity 
    
    @property
    def quantity(self) -> float:
        return self._quantity
    
    @quantity.setter
    def quantity(self, value: float) -> None:
        value = float(value)
        if value <= 0:
            raise ValueError('Количество должно быть положительным')
        self._quantity = value
    
    def __str__(self) -> str:
        return f'{self.name}: {self.quantity} {self.unit}'
    
    def __repr__(self) -> str:
        return f"Ingredient('{self.name}', {self.quantity}, '{self.unit}')"
    
    def __eq__(self, other: object) -> bool:
        if not isinstance(other, Ingredient):
            return NotImplemented
        return self.name == other.name and self.unit == other.unit
````

#### 1.2. **(2 балла)** Реализация класса `Recipe`

Класс `Recipe` представляет собой рецепт одного блюда – например, пиццы Маргариты. Он нужен для того, чтобы объединить название блюда с набором ингредиентов, которые для него требуются, а также предоставить методы для работы с этими ингредиентами

Этот класс должен иметь следующую структуру:
1. Метод `__init__`, инициализирующий два атрибута:
    - `title` – название блюда, строка
    - `ingredients` – список объектов `Ingredient`
2. Метод `add_ingredient(ingredient: Ingredient)` – добавляет ингредиент в рецепт. Если ингредиент с таким названием и единицей измерения уже есть – увеличивает его количество, а не создает дубликат
3. Статический метод `is_valid_ratio(ratio)` (через `@staticmethod`) – возвращает `True`, если `ratio` является числом и больше нуля, иначе `False`
4. Метод `scale(ratio: float)` – возвращает новый объект `Recipe`, в котором количество каждого ингредиента умножено на `ratio`. Исходный рецепт не изменяется
5. Метод `__len__` – возвращает количество уникальных ингредиентов в рецепте
6. Метод `__str__` – возвращает название блюда и список ингредиентов в читаемом виде


````python
class Recipe:
    def __init__(self, title: str, ingredients: list = None):
        self.title = title
        self.ingredients = ingredients if ingredients is not None else []
    
    def add_ingredient(self, ingredient: Ingredient) -> None:
        for existing in self.ingredients:
            if existing == ingredient:
                existing.quantity += ingredient.quantity
                return
        self.ingredients.append(ingredient)
    
    @staticmethod
    def is_valid_ratio(ratio: float) -> bool:
        try:
            return float(ratio) > 0
        except (TypeError, ValueError):
            return False
    
    def scale(self, ratio: float) -> 'Recipe':
        if not self.is_valid_ratio(ratio):
            raise ValueError("Коэффициент масштабирования должен быть положительным числом")
        
        ratio = float(ratio)
        scaled_ingredients = [
            Ingredient(ing.name, ing.quantity * ratio, ing.unit)
            for ing in self.ingredients
        ]
        return Recipe(self.title, scaled_ingredients)
    
    def __len__(self) -> int:
        return len(self.ingredients)
    
    def __str__(self) -> str:
        lines = [f"Рецепт: {self.title}", "Ингредиенты:"]
        lines.extend(f"  - {ing}" for ing in self.ingredients)
        return "\n".join(lines)
````

#### 1.3. **(1,5 балла)** Реализация класса `ShoppingList`

Реалиуйте класс `ShoppingList`. Он представляет собой список покупок, который включает ингредиенты из одного или нескольких рецептов, и нужен для того, чтобы автоматически суммировать все необходимые продукты перед походом в магазин, учитывая, что разные рецепты могут использовать одни и те же ингредиенты (например, мука нужна и для пиццы Маргарита, и для пиццы 4 Сыра, и для любой классической пиццы)

Этот класс должен иметь следующую структуру:
1. Метод `__init__`, инициализирующий один атрибут:
    - `_items` – внутренний список для хранения ингредиентов с привязкой к рецепту (можно хранить кортежами):
        - `ingredient` – объект класса `Ingredient`
        - `recipe_title` – строка, название рецепта, из которого добавлен этот ингредиент
2. Метод `add_recipe(recipe: Recipe, portions: float)` – добавляет рецепт в список покупок:
    - Если `portions` <= 0 – выбрасывает исключение `ValueError` с сообщением "Количество порций должно быть положительным"
    - Масштабирует рецепт на указанное количество порций (`recipe.scale(portions)`)
    - Для каждого ингредиента из масштабированного рецепта добавляет в `_items` кортеж `(ingredient, recipe.title)`
3. Метод `remove_recipe(title: str)` – удаляет все ингредиенты, относящиеся к рецепту с указанным названием:
    - Удаляет из `_items` все элементы, у которых второй элемент кортежа (`recipe_title`) равен `title`
    - Если рецепта с таким названием нет – ничего не происходит
4. Метод `get_list()` – возвращает итоговый список ингредиентов для покупки. Рекомендуется организовать в виде словаря:
    - Пройти по всем кортежам в `_items`
    - Извлечь `ingredient` из кортежа и создать ключ `(ingredient.name, ingredient.unit)`
    - Если ключ уже есть в словаре – прибавить `ingredient.quantity` к существующему значению
    - Если ключа нет – создать запись с `ingredient.quantity`
    - После обработки всех элементов преобразовать словарь в список объектов `Ingredient` – для каждой пары `(name, unit)`, `quantity` из словаря создать `Ingredient(name, quantity, unit)`
    - Отсортировать полученный список по названию ингредиента и вернуть его
5. Метод `__add__(self, other: ShoppingList)` – объединяет два списка покупок в один и возвращает его. Исходные списки при этом не изменяются


````python
from typing import List, Tuple, Dict
from collections import defaultdict


class ShoppingList:
    
    def __init__(self) -> None:
        self._items: List[Tuple[Ingredient, str]] = []
    
    def add_recipe(self, recipe: Recipe, portions: float) -> None:
        """Добавляет рецепт в список покупок.
        
        Args:
            recipe: объект Recipe
            portions: количество порций (масштабирующий коэффициент)
            
        Raises:
            ValueError: если portions <= 0
        """
        if portions <= 0:
            raise ValueError("Количество порций должно быть положительным")
        
        scaled_recipe = recipe.scale(portions)
        
        for ingredient in scaled_recipe.ingredients:
            self._items.append((ingredient, recipe.title))
    
    def remove_recipe(self, title: str) -> None:
        """Удаляет все ингредиенты рецепта с указанным названием.
        
        Args:
            title: название рецепта для удаления
        """
        self._items = [(ing, recipe_title) for ing, recipe_title in self._items 
                       if recipe_title != title]
    
    def get_list(self) -> List[Ingredient]:
        """Возвращает итоговый список ингредиентов для покупки.
        
        Суммирует одинаковые ингредиенты (по name и unit) и возвращает
        отсортированный по названию список объектов Ingredient.
        
        Returns:
            список объектов Ingredient с суммированными количествами
        """
        quantities: Dict[Tuple[str, str], float] = defaultdict(float)
        
        for ingredient, _ in self._items:
            key = (ingredient.name, ingredient.unit)
            quantities[key] += ingredient.quantity
        
        result = [
            Ingredient(name, quantity, unit)
            for (name, unit), quantity in quantities.items()
        ]
        
        result.sort(key=lambda ing: ing.name)
        
        return result
    
    def __add__(self, other: 'ShoppingList') -> 'ShoppingList':
        """Объединяет два списка покупок.
        
        Args:
            other: другой объект ShoppingList
            
        Returns:
            новый ShoppingList, содержащий ингредиенты из обоих списков
        """
        new_list = ShoppingList()
        
        new_list._items = self._items.copy()
        
        new_list._items.extend(other._items)
        
        return new_list
````

#### 1.4. **(1 балл)** Реализация класса `DietaryRecipe`

Класс `DietaryRecipe` является наследником `Recipe` и представляет рецепт с указанием диетической категории (например, `"веган"`, `"без глютена"`, `"кето"`).

Этот класс должен иметь следующую структуру:

1. Метод `__init__` – принимает `title`, `diet_type` (строка, диетическая категория) и опциональный `ingredients`. Вызывает `__init__` родительского класса через `super()`
2. Переопределённый метод `scale(ratio: float)` – возвращает новый объект **`DietaryRecipe`** (а не `Recipe`) с тем же `diet_type`. Использует `super().scale()` для получения масштабированных ингредиентов
3. Переопределённый метод `__str__` – добавляет к строке родителя префикс с диетической категорией, например: `"[веган] Пицца Маргарита"`


````python
from typing import List, Optional

class DietaryRecipe(Recipe):
    
    def __init__(self, title: str, diet_type: str, ingredients: Optional[List[Ingredient]] = None) -> None:
        """Инициализация диетического рецепта.
        
        Args:
            title: название блюда
            diet_type: диетическая категория (веган, без глютена, кето и т.д.)
            ingredients: список ингредиентов (по умолчанию пустой список)
        """
        super().__init__(title, ingredients)
        self.diet_type = diet_type
    
    def scale(self, ratio: float) -> 'DietaryRecipe':
        """Масштабирует диетический рецепт.
        
        Args:
            ratio: коэффициент масштабирования
            
        Returns:
            Новый объект DietaryRecipe с тем же diet_type и масштабированными ингредиентами
        """
        scaled_recipe = super().scale(ratio)
        
        return DietaryRecipe(
            title=scaled_recipe.title,
            diet_type=self.diet_type,
            ingredients=scaled_recipe.ingredients
        )
    
    def __str__(self) -> str:
        """Человекочитаемое представление диетического рецепта.
        
        Returns:
            Строка вида: "[веган] Пицца Маргарита"
        """
        return f"[{self.diet_type}] {self.title}"
````

````python
import pytest
from ingredient import Ingredient
from recipe import Recipe
from shopping_list import ShoppingList

class TestShoppingList:
    """Тесты для класса ShoppingList."""
    
    def setup_method(self):
        """Подготовка тестовых данных перед каждым тестом."""
        self.margherita_ingredients = [
            Ingredient("Мука", 500, "г"),
            Ingredient("Сыр Моцарелла", 200, "г"),
            Ingredient("Помидоры", 300, "г"),
            Ingredient("Томатный соус", 150, "мл")
        ]
        self.margherita = Recipe("Пицца Маргарита", self.margherita_ingredients)
        
        self.four_cheese_ingredients = [
            Ingredient("Мука", 500, "г"),
            Ingredient("Сыр Пармезан", 100, "г"),
            Ingredient("Сыр Горгонзола", 100, "г"),
            Ingredient("Сыр Фета", 100, "г"),
            Ingredient("Сыр Моцарелла", 100, "г")
        ]
        self.four_cheese = Recipe("Пицца 4 Сыра", self.four_cheese_ingredients)
        
        self.salad_ingredients = [
            Ingredient("Огурцы", 200, "г"),
            Ingredient("Помидоры", 150, "г"),
            Ingredient("Сыр Фета", 100, "г"),
            Ingredient("Оливковое масло", 30, "мл")
        ]
        self.salad = Recipe("Греческий салат", self.salad_ingredients)
    
    def test_add_recipe_single(self):
        """Тест добавления одного рецепта в список покупок."""
        shopping_list = ShoppingList()
        
        shopping_list.add_recipe(self.margherita, 1)
        
        # Проверяем, что в _items добавились ингредиенты (4 штуки)
        assert len(shopping_list._items) == 4
        
        # Проверяем, что каждый элемент - кортеж (Ingredient, recipe_title)
        for ingredient, recipe_title in shopping_list._items:
            assert isinstance(ingredient, Ingredient)
            assert recipe_title == "Пицца Маргарита"
    
    def test_add_recipe_with_portions(self):
        """Тест добавления рецепта с масштабированием на количество порций."""
        shopping_list = ShoppingList()
        
        shopping_list.add_recipe(self.margherita, 2)
        
        # Проверяем, что количество каждого ингредиента удвоено
        quantities = {item[0].name: item[0].quantity for item in shopping_list._items}
        assert quantities["Мука"] == 1000.0  # 500 * 2
        assert quantities["Сыр Моцарелла"] == 400.0  # 200 * 2
        assert quantities["Помидоры"] == 600.0  # 300 * 2
        assert quantities["Томатный соус"] == 300.0  # 150 * 2
    
    def test_add_recipe_multiple_recipes(self):
        """Тест добавления нескольких рецептов в список покупок."""
        shopping_list = ShoppingList()
        
        shopping_list.add_recipe(self.margherita, 1)
        shopping_list.add_recipe(self.four_cheese, 1)
        
        # Всего ингредиентов: 4 + 5 = 9 (без учёта дубликатов в _items)
        assert len(shopping_list._items) == 9
    
    def test_add_recipe_zero_portions_raises_error(self):
        """Тест: добавление с portions = 0 выбрасывает ValueError."""
        shopping_list = ShoppingList()
        
        with pytest.raises(ValueError, match="Количество порций должно быть положительным"):
            shopping_list.add_recipe(self.margherita, 0)
    
    def test_add_recipe_negative_portions_raises_error(self):
        """Тест: добавление с отрицательными portions выбрасывает ValueError."""
        shopping_list = ShoppingList()
        
        with pytest.raises(ValueError, match="Количество порций должно быть положительным"):
            shopping_list.add_recipe(self.margherita, -1)
    
    def test_add_recipe_float_portions(self):
        """Тест: добавление с дробным количеством порций."""
        shopping_list = ShoppingList()
        
        shopping_list.add_recipe(self.margherita, 1.5)
        
        quantities = {item[0].name: item[0].quantity for item in shopping_list._items}
        assert quantities["Мука"] == 750.0
        assert quantities["Сыр Моцарелла"] == 300.0
    
    def test_remove_recipe_existing(self):
        """Тест удаления существующего рецепта."""
        shopping_list = ShoppingList()
        shopping_list.add_recipe(self.margherita, 1)
        shopping_list.add_recipe(self.four_cheese, 1)
        
        assert len(shopping_list._items) == 9
        
        shopping_list.remove_recipe("Пицца Маргарита")
        
        assert len(shopping_list._items) == 5
        for _, recipe_title in shopping_list._items:
            assert recipe_title == "Пицца 4 Сыра"
    
    def test_remove_recipe_removes_all_occurrences(self):
        """Тест: удаление рецепта удаляет все его ингредиенты (даже если добавлен несколько раз)."""
        shopping_list = ShoppingList()
        shopping_list.add_recipe(self.margherita, 1)
        shopping_list.add_recipe(self.margherita, 2)  # добавили ещё раз
        
        assert len(shopping_list._items) == 8 
        
        shopping_list.remove_recipe("Пицца Маргарита")
        
        assert len(shopping_list._items) == 0
    
    def test_remove_recipe_non_existent(self):
        """Тест: удаление несуществующего рецепта не вызывает ошибку."""
        shopping_list = ShoppingList()
        shopping_list.add_recipe(self.margherita, 1)
        
        shopping_list.remove_recipe("Несуществующая пицца")
        
        # Список не изменился
        assert len(shopping_list._items) == 4
    
    def test_remove_recipe_from_empty_list(self):
        """Тест: удаление рецепта из пустого списка не вызывает ошибку."""
        shopping_list = ShoppingList()
        
        shopping_list.remove_recipe("Любой рецепт")
        
        assert len(shopping_list._items) == 0
    
    def test_remove_recipe_partial_match(self):
        """Тест: удаление по точному совпадению названия (не частичному)."""
        shopping_list = ShoppingList()
        shopping_list.add_recipe(self.margherita, 1)
        shopping_list.add_recipe(self.four_cheese, 1)
        
        shopping_list.remove_recipe("Пицца")  # не точное совпадение
        
        # Ничего не удалилось
        assert len(shopping_list._items) == 9
    
    def test_get_list_single_recipe(self):
        """Тест: get_list для одного рецепта возвращает ингредиенты без изменений."""
        shopping_list = ShoppingList()
        shopping_list.add_recipe(self.margherita, 1)
        
        result = shopping_list.get_list()
        
        assert len(result) == 4
        # Проверяем, что все ингредиенты на месте
        ingredients_dict = {ing.name: ing.quantity for ing in result}
        assert ingredients_dict["Мука"] == 500.0
        assert ingredients_dict["Сыр Моцарелла"] == 200.0
        assert ingredients_dict["Помидоры"] == 300.0
        assert ingredients_dict["Томатный соус"] == 150.0
    
    def test_get_list_sums_same_ingredients(self):
        """Тест: одинаковые ингредиенты из разных рецептов суммируются."""
        shopping_list = ShoppingList()
        shopping_list.add_recipe(self.margherita, 1)  # Мука: 500г
        shopping_list.add_recipe(self.four_cheese, 1)  # Мука: 500г
        
        result = shopping_list.get_list()
        
        # Находим муку в результате
        flour = next(ing for ing in result if ing.name == "Мука")
        assert flour.quantity == 1000.0  # 500 + 500
    
    def test_get_list_sums_multiple_occurrences(self):
        """Тест: суммирование ингредиентов, встречающихся в нескольких рецептах."""
        shopping_list = ShoppingList()
        shopping_list.add_recipe(self.margherita, 1)   # Сыр Моцарелла: 200г
        shopping_list.add_recipe(self.four_cheese, 1)  # Сыр Моцарелла: 100г
        
        result = shopping_list.get_list()
        
        mozzarella = next(ing for ing in result if ing.name == "Сыр Моцарелла")
        assert mozzarella.quantity == 300.0
    
    def test_get_list_no_duplicates(self):
        """Тест: get_list не содержит дубликатов ингредиентов."""
        shopping_list = ShoppingList()
        shopping_list.add_recipe(self.margherita, 1)
        shopping_list.add_recipe(self.four_cheese, 1)
        shopping_list.add_recipe(self.salad, 1)
        
        result = shipping_list.get_list()
        
        # Проверяем, что нет двух ингредиентов с одинаковыми name и unit
        seen = set()
        for ing in result:
            key = (ing.name, ing.unit)
            assert key not in seen
            seen.add(key)
    
    def test_get_list_sorted_by_name(self):
        """Тест: get_list возвращает список, отсортированный по названию ингредиента."""
        shopping_list = ShoppingList()
        shopping_list.add_recipe(self.four_cheese, 1)
        
        result = shopping_list.get_list()
        
        # Ожидаемый порядок по алфавиту
        names = [ing.name for ing in result]
        assert names == sorted(names)
    
    def test_get_list_sorted_with_multiple_recipes(self):
        """Тест: сортировка с несколькими рецептами."""
        shopping_list = ShoppingList()
        shopping_list.add_recipe(self.margherita, 1)
        shopping_list.add_recipe(self.four_cheese, 1)
        shopping_list.add_recipe(self.salad, 1)
        
        result = shopping_list.get_list()
        
        names = [ing.name for ing in result]
        assert names == sorted(names)
    
    def test_get_list_empty(self):
        """Тест: get_list для пустого списка возвращает пустой список."""
        shopping_list = ShoppingList()
        
        result = shopping_list.get_list()
        
        assert result == []
    
    def test_get_list_preserves_units(self):
        """Тест: get_list правильно различает ингредиенты по единицам измерения."""
        shopping_list = ShoppingList()
        # Добавляем муку в граммах и килограммах
        recipe1 = Recipe("Рецепт 1", [Ingredient("Мука", 500, "г")])
        recipe2 = Recipe("Рецепт 2", [Ingredient("Мука", 1, "кг")])
        
        shopping_list.add_recipe(recipe1, 1)
        shopping_list.add_recipe(recipe2, 1)
        
        result = shopping_list.get_list()
        
        # Должно быть два разных ингредиента
        assert len(result) == 2
        flour_grams = next(ing for ing in result if ing.unit == "г")
        flour_kg = next(ing for ing in result if ing.unit == "кг")
        assert flour_grams.quantity == 500.0
        assert flour_kg.quantity == 1.0
    
    def test_add_two_shopping_lists(self):
        """Тест: объединение двух списков покупок."""
        list1 = ShoppingList()
        list2 = ShoppingList()
        
        list1.add_recipe(self.margherita, 1)
        list2.add_recipe(self.four_cheese, 1)
        
        combined = list1 + list2
        
        assert len(combined._items) == 9
    
    def test_add_does_not_modify_original_lists(self):
        """Тест: исходные списки не изменяются при объединении."""
        list1 = ShoppingList()
        list2 = ShoppingList()
        
        list1.add_recipe(self.margherita, 1)
        list2.add_recipe(self.four_cheese, 1)
        
        original_len1 = len(list1._items)
        original_len2 = len(list2._items)
        
        combined = list1 + list2
        
        assert len(list1._items) == original_len1
        assert len(list2._items) == original_len2
    
    def test_add_returns_new_object(self):
        """Тест: __add__ возвращает новый объект ShoppingList."""
        list1 = ShoppingList()
        list2 = ShoppingList()
        
        list1.add_recipe(self.margherita, 1)
        list2.add_recipe(self.four_cheese, 1)
        
        combined = list1 + list2
        
        assert combined is not list1
        assert combined is not list2
    
    def test_add_with_empty_list(self):
        """Тест: объединение с пустым списком."""
        list1 = ShoppingList()
        list2 = ShoppingList()
        
        list1.add_recipe(self.margherita, 1)
        
        combined = list1 + list2
        
        assert len(combined._items) == len(list1._items)
        assert combined._items == list1._items
    
    def test_add_with_same_recipe_in_both_lists(self):
        """Тест: объединение списков с одинаковыми рецептами."""
        list1 = ShoppingList()
        list2 = ShoppingList()
        
        list1.add_recipe(self.margherita, 1)
        list2.add_recipe(self.margherita, 2)
        
        combined = list1 + list2
        
        # Должно быть 4 + 4 = 8 элементов (дубликаты не суммируются на уровне _items)
        assert len(combined._items) == 8
    
    def test_add_preserves_recipe_titles(self):
        """Тест: при объединении сохраняется привязка к рецептам."""
        list1 = ShoppingList()
        list2 = ShoppingList()
        
        list1.add_recipe(self.margherita, 1)
        list2.add_recipe(self.four_cheese, 1)
        
        combined = list1 + list2
        
        titles = {recipe_title for _, recipe_title in combined._items}
        assert "Пицца Маргарита" in titles
        assert "Пицца 4 Сыра" in titles


if __name__ == "__main__":
    pytest.main([__file__, "-v"])
````
